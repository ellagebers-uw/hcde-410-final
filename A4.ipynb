**Motivation & problem — why this analysis, why it's interesting or useful, what you hope to learn.**
From clicking through the Feeding America food insecurity map, it is clear that food insecurity has grown in seemingly all Washington counties in just four years (2020-2024). The USDA has stopped collecting food insecurity data, which means the data we currently have is a valuable resource to understanding current trends. I want to understand how insecurity rates in Washington have changed since 2020, and what demographics are most affected.

**Data — the dataset(s), what they contain, a link, the license/terms, why they fit, and any ethical considerations.**
The dataset I will use is from Feeding America and their study *Map the Meal Gap.* The project is a tool for Feeding America and partner organizations to understand food insecurity and inform policy decisions. The data contains information on food insecurity rates, SNAP program elegibility, average meal costs, and food budget shortfalls down to country, state, and county levels. The map also contains demographic data. 

The interactive map is linked here: https://map.feedingamerica.org/county/2024/overall/washington

I requested the full dataset, which can be accesed here: https://www.feedingamerica.org/research/map-the-meal-gap/by-county

The dataset is publicly available and can be used for research, non-commercial, and educational use, with the “hope that the nationwide network of food banks, statewide state associations, food pantries and meal programs, as well as people facing food insecurity, policymakers, organizations and supporters will use this information and engage in the movement to end hunger.”
