# Background / related work — what's already known, with references, and how it shapes your study.

Food insecurity is a reduced quality and variability of diet with potential disruptions of eating patterns and food intake due to a lack of money or access to food resources. Food insecurity is a “household-level economic and social condition of limited or uncertain access to adequate food” [4].  Food insecurity is a systemic issue stemming from income-related factors, high cost of living, community factors, health-related factors, and systemic barriers to opportunities. Food insecurity can cause serious effects on people’s physical and mental health, wellbeing, and school and work life [2].

SNAP is the largest anti-hunger program in the United States and is funded by the federal government. States distribute benefits to residents if they meet eligibility standards. SNAP benefits can only be used on food, helping families budget for other expenses. The majority of families spend their benefits immediately, showcasing the importance of these monthly benefits [1]. The Trump administration has reshaped SNAP eligibility categories and the benefits received by households. On November 10th 2026, more regulations were proposed which may eliminate 6.3 million more people from SNAP in the 2029 fiscal year, and remove nearly 1.8 million children from free school meals [3].

I’ve decided to include SNAP benefit data in my study since SNAP is currently undergoing major cuts. While the past data may not reflect current situations, I think it is important to highlight how many food insecure individuals live in households with income above SNAP thresholds. With large cuts being imposed, there may be even more individuals facing food insecurity who once met SNAP criteria. I want my project to highlight how food insecurity is changing overtime while also highlighting the importance of food assistance resources.

# Methodology — the analytical methods you'll use and why they fit, and how you'll present your findings.
Currently, the data visualization Map the Meal Gap makes comparing food insecurity between counties in a specific year easy to do. However, it is hard to see how food insecurity has changed on a county level over time. I will clean up the dataset from the Map the Meal Gap data for the years 2018-2024, specifically looking at the Washington State data at the county level. 

I will include:
- Overall Food Insecurity Rate
- Food Insecure Population
- The percentage of food insecure individuals who live in households with income at or below SNAP threshold in the state
- The percentage of food insecure individuals who live in households with income above SNAP threshold in the state
- Cost per meal - The average dollar amount spent on food per meal by food-secure individuals
- Annual Budget Shortfall

Then I will create a data visualization that will allow the user to easily see how food insecurity has changed over time across counties. Presenting my findings this way will help users understand how food insecurity has changed over time, and what this means for the future if we continue following trends.

The *Note on Compatibility of Local Food Insecurity Estimates* published by Feeding America states: “... county and service area food insecurity estimates from Map the Meal Gap 2023 (2021 data) may be compared to data from Map the Meal Gap 2022 (2020 data), Map the Meal Gap 2021 (2019 data), and Map the Meal Gap 2020 (2018 data). We do not recommend comparing food insecurity estimates for any geography from the most recent four studies – Map the Meal Gap 2023 (2021 data), Map the Meal Gap 2022 (2020 data), Map the Meal Gap 2021 (2019 data), or Map the Meal Gap 2020 (2018 data) – to estimates from Map the Meal Gap 2019 (2017 data) or any previous year due to the changes in the methodology made in 2020 (i.e., updated poverty variable and new disability variable). Estimates from Map the Meal Gap 2013 (2011 data) through Map the Meal Gap 2019 (2017 data) are more directly comparable.”

To keep data collection methodology consistent the data I use from this project will be for the years 2018-2024, and not include data collected before 2018.
